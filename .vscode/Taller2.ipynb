Objetivo 1 - graficar la función
Vector xs
Definir función f(x)
Calcular vector f(xs) = ys

```python
def f(x):
    
    y = x**3 - 3 * x**2 + x - 1
    print(x,y)
    return y
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import newton

# ==========================================
# FUNCION Y DERIVADA
# ==========================================

def f(x):
    return x**3 - 3*x**2 + x - 1

def df(x):
    return (3*x**2) - (6*x) + 1

# ==========================================
# METODO DE NEWTON
# ==========================================

# Aproximación inicial
x0 = 100

# Calcular raíz
raiz = newton(f, x0=x0, fprime=df)

# ==========================================
# DATOS PARA GRAFICAR
# ==========================================

x = np.linspace(-2, 4, 1000)

y_funcion = f(x)
y_derivada = df(x)

# ==========================================
# GRAFICAS
# ==========================================

plt.figure(figsize=(12, 7))

# Función original
plt.plot(x, y_funcion, label='f(x) = x³ - 3x² + x - 1')

# Derivada
plt.plot(x, y_derivada, label="f'(x) = 3x² - 6x + 1")

# Eje X
plt.axhline(y=0)

# Eje Y
plt.axvline(x=0)

# Línea vertical en la raíz
plt.axvline(x=raiz, linestyle='--', label=f'Raíz ≈ {raiz:.6f}')

# Punto raíz
plt.scatter(raiz, f(raiz), s=100)

# Cuadrícula
plt.grid()

# Límites
plt.xlim(-2, 4)
plt.ylim(-15, 15)

# Título
plt.title('Método de Newton - Función y Derivada')

# Etiquetas
plt.xlabel('x')
plt.ylabel('y')

# Leyenda
plt.legend()

# Mostrar gráfica
plt.show()

# ==========================================
# RESULTADOS
# ==========================================

print("===================================")
print(" MÉTODO DE NEWTON ")
print("===================================")

print(f"Aproximación inicial x0 = {x0}")
print(f"Raíz encontrada = {raiz}")
print(f"f(raiz) = {f(raiz)}")
```

```python
import numpy as np

xs = list(range(-2, 6))
xs = list(np.linspace(-2, 6, 100))
```

```python
def f(x):
    _y = x**3 - 3 * x**2 + x - 1
    return _y
```

```python
ys = [f(x) for x in xs]
```

```python
import matplotlib.pyplot as plt

plt.plot(xs, ys, ".:")
plt.xlabel("xs")
plt.ylabel("ys")
plt.title("Gráfica de ys vs xs")
plt.grid(True)
plt.show()
```

```python
from scipy.optimize import newton

# 1. Creamos listas normales de Python para guardar los datos
valores_x = []
valores_fx = []

# 2. Definimos la función y capturamos los datos en cada evaluación de SciPy
def f(x):
    y = x**3 - 3*x**2 + x - 1
    valores_x.append(x)
    valores_fx.append(y)
    return y

# 3. Ejecutamos el método de Newton puro de SciPy
raiz_final = newton(f, x0=100)

# 4. Presentamos el resultado en dos columnas usando formato de texto clásico
print(f"{'x_n':<25} | {'f(x_n)':<25}")
print("-" * 53)

# Emparejamos los datos de ambas listas y los imprimimos fila por fila
for x, fx in zip(valores_x, valores_fx):
    print(f"{x:<25.10f} | {fx:<25.10f}")

print("-" * 53)
print(f"Raíz final encontrada: {raiz_final}")
```
