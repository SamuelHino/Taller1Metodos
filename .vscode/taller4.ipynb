```python
# Comparación con el valor de cos(1), simplifcado el polinomio a orden 0, es decir, p0 = cos(1)
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3,3,1000)

y = np.cos(x)

p0 = np.cos(1)

plt.plot(x,y,label="cos(x)")
plt.plot(x,p0*np.ones_like(x),label="P0")

plt.grid(True)
plt.legend()
plt.show()
```

```python
# Para caso inical x0 = 1
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 1000)

# Función original
y = np.cos(x)

y0 = 1
# y1 = 1 El orden 1 coincide con el orden 0

# El orden impar resulta exactamente igual al orden par anterior.
y2 = 1 - x**2 / 2
# y3 = 1 - x**2/ 2 + 0 * x**3 / 6
y4 = 1- x**2 / 2 + x**4 / 24


plt.figure(figsize=(10, 6))

plt.plot(x, y, label='cos(x)', linewidth=3)

plt.plot(x, y0*np.ones_like(x), label='Orden 0')
# plt.plot(x, y1*np.ones_like(x), '--', label='Orden 1')

plt.plot(x, y2, label='Orden 2')
# plt.plot(x, y3, '--', label='Orden 3')

plt.plot(x, y4, label='Orden 4')

plt.grid(True)
plt.legend()
plt.title('Serie de Taylor de cos(x) alrededor de x₀ = 0')

plt.show()

```

```python
# Para caso inical x0 = 1
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 1000)

# Función original
y = np.cos(x)

# Taylor alrededor de x0 = 1

y0 = np.cos(1)

y1 = np.cos(1) - np.sin(1)*(x - 1)

y2 = (
    np.cos(1)
    - np.sin(1)*(x - 1)
    - np.cos(1)*(x - 1)**2 / 2
)

y3 = (
    np.cos(1)
    - np.sin(1)*(x - 1)
    - np.cos(1)*(x - 1)**2 / 2
    + np.sin(1)*(x - 1)**3 / 6
)

y4 = (
    np.cos(1)
    - np.sin(1)*(x - 1)
    - np.cos(1)*(x - 1)**2 / 2
    + np.sin(1)*(x - 1)**3 / 6
    + np.cos(1)*(x - 1)**4 / 24
)

plt.figure(figsize=(10, 6))

# Función base punteada
plt.plot(
    x,
    y,
    '--',
    linewidth=2,
    label='Función base: cos(x)'
)

plt.plot(x, y0*np.ones_like(x), label='Orden 0')
plt.plot(x, y1, label='Orden 1')
plt.plot(x, y2, label='Orden 2')
plt.plot(x, y3, label='Orden 3')
plt.plot(x, y4, label='Orden 4')

plt.grid(True)
plt.legend()
plt.title('Serie de Taylor de cos(x) alrededor de x₀ = 1')

plt.show()
```

```python
# Para caso inical x0 = 0 con la formula 1/1-x
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-3, 3, 1000)
y = 1/(1-x)

p0 = 1/(1-0)
plt.plot(x,y,label="1/(1-x)")
plt.plot(x,p0*np.ones_like(x),label="P0")
plt.grid(True)
plt.legend()
```
