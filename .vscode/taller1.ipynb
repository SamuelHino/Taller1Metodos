```python
import matplotlib.pyplot as plt

# Lista de tokens
tokens = list(range(0, 1_000_000, 1000))

# Lista de precios
precio = []

# Punto donde cambia la pendiente
limite = 300_000

for t in tokens:

    # Primera recta
    if t < limite:

        # Va creciendo lentamente hasta llegar a 45
        p = (45 / limite) * t

    # Segunda recta
    else:

        # Continúa desde 45 pero con mayor pendiente
        p = 45 + 0.0002 * (t - limite)

    precio.append(p)

# Graficar
plt.figure(figsize=(10,5))

plt.plot(tokens, precio, label="Precio vs Tokens")

# Línea vertical de separación
plt.axvline(x=limite, linestyle='--', label='Cambio de pendiente')

# Línea horizontal en y = 45
# plt.axhline(y=45, linestyle='--')

# Etiquetas
plt.xlabel("Tokens")
plt.ylabel("Precio ($)")
plt.title("Costo de Tokens en IA")

plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


# Función
def f(x):
    return (1/4) * (x**3 + 3*x**2 - 6*x - 8)


# Método de Bisección
def biseccion(a, b, iteraciones=5):

    print(f"\nIntervalo inicial: [{a}, {b}]")

    # Verificar si existe raíz en el intervalo
    if f(a) * f(b) > 0:
        print("No hay cambio de signo en este intervalo.")
        return None

    for i in range(iteraciones):

        # Punto medio
        c = (a + b) / 2

        print(f"Iteración {i+1}")
        print(f"a = {a:.6f}")
        print(f"b = {b:.6f}")
        print(f"c = {c:.6f}")
        print(f"f(c) = {f(c):.6f}")
        print("-" * 30)

        # Verificar en qué subintervalo está la raíz
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    return c

# Rangos pedidos
rangos = [
    (-5, 3),
    (-4.7, 2.5),
    (-4.5, 2.7),
    (-3, -2)
]

# Ejecutar bisección
raices = []

for a, b in rangos:
    raiz = biseccion(a, b, iteraciones=5)
    raices.append(raiz)


# Graficar función
x = np.linspace(-6, 4, 1000)
y = f(x)

plt.figure(figsize=(10, 6))

# Función
plt.plot(x, y, label='f(x)')

# Eje X
plt.axhline(0, color='black')

# Marcar raíces aproximadas
for r in raices:
    if r is not None:
        plt.plot(r, f(r), marker='o')

# Título y etiquetas
plt.title('Método de Bisección')
plt.xlabel('x')
plt.ylabel('f(x)')

# Cuadrícula
plt.grid(True)

# Leyenda
plt.legend()

# Mostrar gráfica
plt.show()
```
