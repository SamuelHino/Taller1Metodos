## Interpolación por minimos cuadrados

* Parte 1 Comprobar ejerccios

```python
# Comprobación 1er ejercicio
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# DATOS
# ==========================================================

xs = np.array([0, 0, 2, 2])
ys = np.array([1, 0, 2, 1])

# ==========================================================
# GRADO DEL POLINOMIO
# ==========================================================
# 1 -> lineal
# 2 -> cuadrático
# 3 -> cúbico

grado = 1

# ==========================================================
# AJUSTE POR MÍNIMOS CUADRADOS
# ==========================================================

coeficientes = np.polyfit(xs, ys, grado)

# Crear función polinómica
funcion = np.poly1d(coeficientes)

# ==========================================================
# MOSTRAR ECUACIÓN
# ==========================================================

print("Coeficientes encontrados:")
print(coeficientes)

print("\nFunción aproximada:")
print(funcion)

# ==========================================================
# CALCULAR PREDICCIONES
# ==========================================================

ys_predichos = funcion(xs)

print("\nValores predichos:")

for x, y_real, y_pred in zip(xs, ys, ys_predichos):

    error = abs(y_real - y_pred)

    print(f"x={x}")
    print(f"Real      = {y_real}")
    print(f"Predicho  = {y_pred:.4f}")
    print(f"Error     = {error:.4f}")
    print("--------------------")

# ==========================================================
# GRAFICAR
# ==========================================================

x_grafica = np.linspace(min(xs)-1, max(xs)+1, 500)
y_grafica = funcion(x_grafica)

plt.figure(figsize=(8,5))

# Puntos originales
plt.scatter(xs, ys, s=100, label="Datos")

# Curva ajustada
plt.plot(x_grafica, y_grafica, label="Mínimos cuadrados")

plt.grid(True)

plt.xlabel("x")
plt.ylabel("y")

plt.title("Ajuste por mínimos cuadrados")

plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt



def aproximacion_minimos_cuadrados(xs, ys, grado):
    """
    Ajuste por mínimos cuadrados.

    Parámetros:
    xs -> lista de valores x
    ys -> lista de valores y
    grado -> grado del polinomio:
              1 = lineal
              2 = cuadrático
              3 = cúbico
              etc.
    """

    # ======================================================
    # Convertir a arrays
    # ======================================================

    xs = np.array(xs)
    ys = np.array(ys)

    # ======================================================
    # GRAFICAR DATOS ORIGINALES
    # ======================================================

    plt.figure(figsize=(8,5))

    plt.scatter(xs, ys, s=100, label="Datos originales")

    plt.title("Datos originales")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True)

    plt.legend()

    plt.show()

    # ======================================================
    # AJUSTE POR MÍNIMOS CUADRADOS
    # ======================================================

    coeficientes = np.polyfit(xs, ys, grado)

    # Crear polinomio
    funcion = np.poly1d(coeficientes)

    # ======================================================
    # MOSTRAR FUNCIÓN
    # ======================================================

    print("\n===================================")
    print("FUNCIÓN DE APROXIMACIÓN")
    print("===================================\n")

    print(funcion)

    # ======================================================
    # CALCULAR ERRORES
    # ======================================================

    ys_pred = funcion(xs)

    print("\n===================================")
    print("ERRORES")
    print("===================================\n")

    for x, y_real, y_aprox in zip(xs, ys, ys_pred):

        error = abs(y_real - y_aprox)

        print(f"x = {x}")
        print(f"Real      = {y_real}")
        print(f"Aproximado= {y_aprox:.6f}")
        print(f"Error     = {error:.6f}")
        print("---------------------------")

    # ======================================================
    # GRAFICAR APROXIMACIÓN
    # ======================================================

    x_graf = np.linspace(min(xs)-1, max(xs)+1, 500)

    y_graf = funcion(x_graf)

    plt.figure(figsize=(10,6))

    # puntos originales
    plt.scatter(xs, ys, s=100, label="Datos")

    # curva aproximada
    plt.plot(x_graf, y_graf, linewidth=2,
             label=f"Polinomio grado {grado}")

    plt.title("Ajuste por mínimos cuadrados")

    plt.xlabel("x")
    plt.ylabel("y")

    plt.grid(True)

    plt.legend()

    plt.show()

    return funcion
```

```python
import numpy as np
import matplotlib.pyplot as plt

def minimos_cuadrados_lineal(xs, ys):
    """
    Ajuste lineal: y = a0 + a1*x
    """
    n = len(xs)

    sum_x   = 0
    sum_y   = 0
    sum_x2  = 0
    sum_xy  = 0

    for x, y in zip(xs, ys):
        sum_x  += x
        sum_y  += y
        sum_x2 += x**2
        sum_xy += x * y

    A = [
        [n,      sum_x],
        [sum_x,  sum_x2]
    ]
    b = [sum_y, sum_xy]

    coef = np.linalg.solve(A, b)
    a0, a1 = coef

    x_grafica = np.linspace(min(xs)-1, max(xs)+1, 500)
    y_grafica = a0 + a1 * x_grafica

    plt.figure(figsize=(8, 5))
    plt.scatter(xs, ys, s=100, label="Datos")
    plt.plot(x_grafica, y_grafica, color="red",
             label=f"y = {a1:.2f}x + {a0:.2f}")
    plt.grid(True)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Mínimos cuadrados — Lineal")
    plt.legend()
    plt.show()

    print(f"=== Lineal ===")
    print(f"  a0 (b) = {a0:.4f}")
    print(f"  a1 (m) = {a1:.4f}")
    print(f"  Ecuación: y = {a1:.4f}x + {a0:.4f}")

    return a0, a1

# Lista de puntos de ejemplo
xs = [3.38, 0.35, 2.07, -0.45, -0.55, -1.06, -2.77, 3.08,
      -3.98, -5., -3.18, -1.96, 2.37, -1.66, 4.09]
ys = [-0.04, 1.28, 0.65, 2.18, 1.70, 1.96, 3.36, 0.18,
       3.35, 4.16, 2.95, 2.34, 0.38, 1.75, -1.06]

minimos_cuadrados_lineal(xs, ys)
        
```

```python
def minimos_cuadrados_cuadratico(xs, ys):
    """
    Ajuste cuadrático: y = a0 + a1*x + a2*x²
    """
    n = len(xs)

    sum_x   = 0
    sum_y   = 0
    sum_x2  = 0
    sum_x3  = 0
    sum_x4  = 0
    sum_xy  = 0
    sum_x2y = 0

    for x, y in zip(xs, ys):
        sum_x   += x
        sum_y   += y
        sum_x2  += x**2
        sum_x3  += x**3
        sum_x4  += x**4
        sum_xy  += x * y
        sum_x2y += x**2 * y

    A = [
        [n,       sum_x,   sum_x2],
        [sum_x,   sum_x2,  sum_x3],
        [sum_x2,  sum_x3,  sum_x4]
    ]
    b = [sum_y, sum_xy, sum_x2y]

    coef = np.linalg.solve(A, b)
    a0, a1, a2 = coef

    x_grafica = np.linspace(min(xs)-1, max(xs)+1, 500)
    y_grafica = a0 + a1*x_grafica + a2*x_grafica**2

    plt.figure(figsize=(8, 5))
    plt.scatter(xs, ys, s=100, label="Datos")
    plt.plot(x_grafica, y_grafica, color="red",
             label=f"y = {a2:.2f}x² + {a1:.2f}x + {a0:.2f}")
    plt.grid(True)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Mínimos cuadrados — Cuadrático")
    plt.legend()
    plt.show()

    print(f"=== Cuadrático ===")
    print(f"  a0 = {a0:.4f}")
    print(f"  a1 = {a1:.4f}")
    print(f"  a2 = {a2:.4f}")
    print(f"  Ecuación: y = {a2:.4f}x² + {a1:.4f}x + {a0:.4f}")

    return a0, a1, a2

minimos_cuadrados_cuadratico(xs, ys)
```

```python
def minimos_cuadrados_cubico(xs, ys):
    """
    Ajuste cúbico: y = a0 + a1*x + a2*x² + a3*x³
    """
    n = len(xs)

    sum_x   = 0
    sum_y   = 0
    sum_x2  = 0
    sum_x3  = 0
    sum_x4  = 0
    sum_x5  = 0
    sum_x6  = 0
    sum_xy  = 0
    sum_x2y = 0
    sum_x3y = 0

    for x, y in zip(xs, ys):
        sum_x   += x
        sum_y   += y
        sum_x2  += x**2
        sum_x3  += x**3
        sum_x4  += x**4
        sum_x5  += x**5
        sum_x6  += x**6
        sum_xy  += x * y
        sum_x2y += x**2 * y
        sum_x3y += x**3 * y

    A = [
        [n,       sum_x,   sum_x2,  sum_x3],
        [sum_x,   sum_x2,  sum_x3,  sum_x4],
        [sum_x2,  sum_x3,  sum_x4,  sum_x5],
        [sum_x3,  sum_x4,  sum_x5,  sum_x6]
    ]
    b = [sum_y, sum_xy, sum_x2y, sum_x3y]

    coef = np.linalg.solve(A, b)
    a0, a1, a2, a3 = coef

    x_grafica = np.linspace(min(xs)-1, max(xs)+1, 500)
    y_grafica = a0 + a1*x_grafica + a2*x_grafica**2 + a3*x_grafica**3

    plt.figure(figsize=(8, 5))
    plt.scatter(xs, ys, s=100, label="Datos")
    plt.plot(x_grafica, y_grafica, color="red",
             label=f"y = {a3:.2f}x³ + {a2:.2f}x² + {a1:.2f}x + {a0:.2f}")
    plt.grid(True)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Mínimos cuadrados — Cúbico")
    plt.legend()
    plt.show()

    print(f"=== Cúbico ===")
    print(f"  a0 = {a0:.4f}")
    print(f"  a1 = {a1:.4f}")
    print(f"  a2 = {a2:.4f}")
    print(f"  a3 = {a3:.4f}")
    print(f"  Ecuación: y = {a3:.4f}x³ + {a2:.4f}x² + {a1:.4f}x + {a0:.4f}")

    return a0, a1, a2, a3

minimos_cuadrados_cubico(xs, ys)
```

### Parte 3

- Calcular el error absoluto promedio y su desviación estándar. 
- Encontrar el punto con el error máximo y mínimo. Marcar estos puntos en el gráfico.  

```python
import numpy as np
import matplotlib.pyplot as plt

xs = [3.38, 0.35, 2.07, -0.45, -0.56, -1.06, -2.78, 3.08,
      -3.99, -5.00, -3.18, -1.97, 2.37, -1.67, 4.09,
      -4.60, 2.68, 2.78, -3.79, -1.87]
ys = [15.65, -11.20, -4.00, 18.03, 7.94, 15.32, -4.40, 1.39,
      -11.92, -90.24, 6.92, 28.35, -2.41, -5.47, 35.91,
      -55.53, 0.77, 4.79, -27.05, 2.85]


def analizar_errores(xs, ys, y_pred, titulo):
    errores = [abs(ys[i] - y_pred[i]) for i in range(len(ys))]

    error_promedio = sum(errores) / len(errores)
    varianza = sum((e - error_promedio)**2 for e in errores) / len(errores)
    desviacion = varianza**0.5

    idx_max = errores.index(max(errores))
    idx_min = errores.index(min(errores))

    print(f"\n  Error absoluto promedio : {error_promedio:.4f}")
    print(f"  Desviación estándar     : {desviacion:.4f}")
    print(f"  Error máximo            : {errores[idx_max]:.4f}  en x={xs[idx_max]}, y={ys[idx_max]}")
    print(f"  Error mínimo            : {errores[idx_min]:.4f}  en x={xs[idx_min]}, y={ys[idx_min]}")

    return idx_max, idx_min


def minimos_cuadrados_lineal(xs, ys):
    n = len(xs)

    sum_x  = 0
    sum_y  = 0
    sum_x2 = 0
    sum_xy = 0

    for x, y in zip(xs, ys):
        sum_x  += x
        sum_y  += y
        sum_x2 += x**2
        sum_xy += x * y

    A = [
        [n,     sum_x],
        [sum_x, sum_x2]
    ]
    b = [sum_y, sum_xy]

    coef = np.linalg.solve(A, b)
    a0, a1 = coef

    x_grafica = np.linspace(min(xs)-1, max(xs)+1, 500)
    y_grafica = a0 + a1 * x_grafica
    y_pred    = [a0 + a1 * x for x in xs]

    print(f"=== Lineal ===")
    print(f"  Ecuación: y = {a1:.4f}x + {a0:.4f}")
    idx_max, idx_min = analizar_errores(xs, ys, y_pred, "Lineal")

    plt.figure(figsize=(8, 5))
    plt.scatter(xs, ys, s=80, color="blue", label="Datos", zorder=5)
    plt.plot(x_grafica, y_grafica, color="red",
             label=f"y = {a1:.2f}x + {a0:.2f}")
    plt.scatter(xs[idx_max], ys[idx_max], s=150, color="red",
                zorder=6, label=f"Error máx ({xs[idx_max]:.2f}, {ys[idx_max]:.2f})")
    plt.scatter(xs[idx_min], ys[idx_min], s=150, color="green",
                zorder=6, label=f"Error mín ({xs[idx_min]:.2f}, {ys[idx_min]:.2f})")
    plt.grid(True)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Mínimos cuadrados — Lineal")
    plt.legend()
    plt.show()

    return a0, a1


def minimos_cuadrados_cuadratico(xs, ys):
    n = len(xs)

    sum_x   = 0
    sum_y   = 0
    sum_x2  = 0
    sum_x3  = 0
    sum_x4  = 0
    sum_xy  = 0
    sum_x2y = 0

    for x, y in zip(xs, ys):
        sum_x   += x
        sum_y   += y
        sum_x2  += x**2
        sum_x3  += x**3
        sum_x4  += x**4
        sum_xy  += x * y
        sum_x2y += x**2 * y

    A = [
        [n,      sum_x,  sum_x2],
        [sum_x,  sum_x2, sum_x3],
        [sum_x2, sum_x3, sum_x4]
    ]
    b = [sum_y, sum_xy, sum_x2y]

    coef = np.linalg.solve(A, b)
    a0, a1, a2 = coef

    x_grafica = np.linspace(min(xs)-1, max(xs)+1, 500)
    y_grafica = a0 + a1*x_grafica + a2*x_grafica**2
    y_pred    = [a0 + a1*x + a2*x**2 for x in xs]

    print(f"=== Cuadrático ===")
    print(f"  Ecuación: y = {a2:.4f}x² + {a1:.4f}x + {a0:.4f}")
    idx_max, idx_min = analizar_errores(xs, ys, y_pred, "Cuadrático")

    plt.figure(figsize=(8, 5))
    plt.scatter(xs, ys, s=80, color="blue", label="Datos", zorder=5)
    plt.plot(x_grafica, y_grafica, color="red",
             label=f"y = {a2:.2f}x² + {a1:.2f}x + {a0:.2f}")
    plt.scatter(xs[idx_max], ys[idx_max], s=150, color="red",
                zorder=6, label=f"Error máx ({xs[idx_max]:.2f}, {ys[idx_max]:.2f})")
    plt.scatter(xs[idx_min], ys[idx_min], s=150, color="green",
                zorder=6, label=f"Error mín ({xs[idx_min]:.2f}, {ys[idx_min]:.2f})")
    plt.grid(True)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Mínimos cuadrados — Cuadrático")
    plt.legend()
    plt.show()

    return a0, a1, a2


def minimos_cuadrados_cubico(xs, ys):
    n = len(xs)

    sum_x   = 0
    sum_y   = 0
    sum_x2  = 0
    sum_x3  = 0
    sum_x4  = 0
    sum_x5  = 0
    sum_x6  = 0
    sum_xy  = 0
    sum_x2y = 0
    sum_x3y = 0

    for x, y in zip(xs, ys):
        sum_x   += x
        sum_y   += y
        sum_x2  += x**2
        sum_x3  += x**3
        sum_x4  += x**4
        sum_x5  += x**5
        sum_x6  += x**6
        sum_xy  += x * y
        sum_x2y += x**2 * y
        sum_x3y += x**3 * y

    A = [
        [n,      sum_x,  sum_x2, sum_x3],
        [sum_x,  sum_x2, sum_x3, sum_x4],
        [sum_x2, sum_x3, sum_x4, sum_x5],
        [sum_x3, sum_x4, sum_x5, sum_x6]
    ]
    b = [sum_y, sum_xy, sum_x2y, sum_x3y]

    coef = np.linalg.solve(A, b)
    a0, a1, a2, a3 = coef

    x_grafica = np.linspace(min(xs)-1, max(xs)+1, 500)
    y_grafica = a0 + a1*x_grafica + a2*x_grafica**2 + a3*x_grafica**3
    y_pred    = [a0 + a1*x + a2*x**2 + a3*x**3 for x in xs]

    print(f"=== Cúbico ===")
    print(f"  Ecuación: y = {a3:.4f}x³ + {a2:.4f}x² + {a1:.4f}x + {a0:.4f}")
    idx_max, idx_min = analizar_errores(xs, ys, y_pred, "Cúbico")

    plt.figure(figsize=(8, 5))
    plt.scatter(xs, ys, s=80, color="blue", label="Datos", zorder=5)
    plt.plot(x_grafica, y_grafica, color="red",
             label=f"y = {a3:.2f}x³ + {a2:.2f}x² + {a1:.2f}x + {a0:.2f}")
    plt.scatter(xs[idx_max], ys[idx_max], s=150, color="red",
                zorder=6, label=f"Error máx ({xs[idx_max]:.2f}, {ys[idx_max]:.2f})")
    plt.scatter(xs[idx_min], ys[idx_min], s=150, color="green",
                zorder=6, label=f"Error mín ({xs[idx_min]:.2f}, {ys[idx_min]:.2f})")
    plt.grid(True)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Mínimos cuadrados — Cúbico")
    plt.legend()
    plt.show()

    return a0, a1, a2, a3


# --- Ejecutar ---
minimos_cuadrados_lineal(xs, ys)
minimos_cuadrados_cuadratico(xs, ys)
minimos_cuadrados_cubico(xs, ys)
```
